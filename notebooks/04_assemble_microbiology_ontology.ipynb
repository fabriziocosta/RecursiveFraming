# Assemble a filtered microbiology ontology

Run this notebook before `05_semantic_abstraction.ipynb`. The assembly implementation lives in `graphicalizer.ontology_assembly`; this notebook only sets the two controls and calls it. Assembly starts with the stable generic relation seed, then adds the domain-specific microbiology ontology and filtered external terms.

`abstraction_level=0` favors specific terms, `1` is balanced, and `2` favors broader concepts and ancestors.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / 'assets').exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from graphicalizer import assemble_microbiology_ontology
```

```python
from graphicalizer.notebook_config import load_notebook_config

CONFIG = load_notebook_config(PROJECT_ROOT)
COMMON = CONFIG['common']
SETTINGS = CONFIG['notebook_04_assemble_microbiology_ontology']
TERM_BUDGET = SETTINGS['term_budget']
ABSTRACTION_LEVEL = SETTINGS['abstraction_level']
INCLUDE_NCBITAXON = SETTINGS['include_ncbitaxon']

assembly = assemble_microbiology_ontology(
    PROJECT_ROOT,
    term_budget=TERM_BUDGET,
    abstraction_level=ABSTRACTION_LEVEL,
    include_ncbitaxon=INCLUDE_NCBITAXON,
)

print(f'Selected external terms: {assembly.selected_terms}')
print(f'Abstraction level: {assembly.abstraction_level}')
print(f'Assembled ontology: {assembly.ontology_path}')
print(f'Term catalog: {assembly.catalog_path}')
```

## Prompt-facing ontology

```python
import yaml

assembled = yaml.safe_load(assembly.ontology_path.read_text(encoding='utf-8'))
prompt_ontology = {
    'entity_types': assembled['entity_types'],
    'relation_types': assembled['relation_types'],
}
ontology_yaml = yaml.safe_dump(prompt_ontology, sort_keys=False, allow_unicode=True)
print(ontology_yaml)
```
